```python
from collections import deque

def solve(start, goal):
    q = deque([(start, [])])
    visited = {tuple(start)}

    while q:
        state, path = q.popleft()

        if state == goal:
            for s in path + [state]:
                print(s[:3])
                print(s[3:6])
                print(s[6:])
                print()
            return

        i = state.index(0)
        for j in [i-3, i+3, i-1, i+1]:
            if 0 <= j < 9 and not (abs(i-j) == 1 and i//3 != j//3):
                new = state[:]
                new[i], new[j] = new[j], new[i]
                t = tuple(new)

                if t not in visited:
                    visited.add(t)
                    q.append((new, path + [state]))

start = [1, 2, 3, 4, 0, 6, 7, 5, 8]
goal  = [1, 2, 3, 4, 5, 6, 7, 8, 0]

solve(start, goal)
```

```python
N = 8

def print_solution(board):
    for row in board:
        print(" ".join("Q" if x else "." for x in row))

def is_safe(board, row, col):
    for i in range(row):
        if board[i][col]:
            return False

        if col - (row - i) >= 0:
            if board[i][col - (row - i)]:
                return False

        if col + (row - i) < N:
            if board[i][col + (row - i)]:
                return False

    return True

def solve(board, row):
    if row == N:
        return True

    for col in range(N):
        if is_safe(board, row, col):
            board[row][col] = 1

            if solve(board, row + 1):
                return True

            board[row][col] = 0

    return False

board = [[0] * N for _ in range(N)]

if solve(board, 0):
    print("Solution:")
    print_solution(board)
else:
    print("No solution exists")
```

```python
import heapq

graph = {
    'A': [('B', 1), ('C', 4)],
    'B': [('D', 2), ('E', 5)],
    'C': [('D', 1)],
    'D': [('E', 1)],
    'E': []
}

h = {'A': 5, 'B': 4, 'C': 2, 'D': 1, 'E': 0}

def astar(start, goal):
    q = [(h[start], 0, start, [start])]
    visited = set()

    while q:
        f, g, node, path = heapq.heappop(q)

        if node == goal:
            print("Path:", path)
            print("Cost:", g)
            return

        if node not in visited:
            visited.add(node)

            for nxt, cost in graph[node]:
                if nxt not in visited:
                    new_g = g + cost
                    heapq.heappush(
                        q, (new_g + h[nxt], new_g, nxt, path + [nxt])
                    )

astar('A', 'E')
```

```python
def alphabeta(depth, node, maximizing, alpha, beta, values):
    if depth == 2:
        return values[node]

    if maximizing:
        best = float("-inf")

        for child in [node * 2, node * 2 + 1]:
            value = alphabeta(
                depth + 1, child, False, alpha, beta, values
            )
            best = max(best, value)
            alpha = max(alpha, best)

            if beta <= alpha:
                break

        return best

    else:
        best = float("inf")

        for child in [node * 2, node * 2 + 1]:
            value = alphabeta(
                depth + 1, child, True, alpha, beta, values
            )
            best = min(best, value)
            beta = min(beta, best)

            if beta <= alpha:
                break

        return best

values = [3, 5, 2, 9]

print("Best value:", alphabeta(
    0, 0, True, float("-inf"), float("inf"), values
))
```

```python
import random

def fitness(x):
    return -(x - 5) ** 2 + 25

population = [random.uniform(0, 10) for _ in range(10)]

for generation in range(50):
    population.sort(key=fitness, reverse=True)
    parents = population[:5]

    children = []
    for _ in range(10):
        p1, p2 = random.sample(parents, 2)
        child = (p1 + p2) / 2
        child += random.uniform(-0.5, 0.5)
        child = max(0, min(10, child))
        children.append(child)

    population = parents + children[:5]

best = max(population, key=fitness)

print("Best Solution:", best)
print("Maximum Fitness:", fitness(best))
```

```python
colors = ["Red", "Green", "Blue"]

graph = {
    "A": ["B", "C"],
    "B": ["A", "C", "D"],
    "C": ["A", "B", "D"],
    "D": ["B", "C"]
}

result = {}

def color_map(node):
    if node == len(graph):
        return True

    region = list(graph.keys())[node]

    for color in colors:
        if all(result.get(n) != color for n in graph[region]):
            result[region] = color

            if color_map(node + 1):
                return True

            del result[region]

    return False

if color_map(0):
    print("Map Coloring:")
    print(result)
else:
    print("No solution")
```

```python
def minimax(depth, node, maximizing, values):
    if depth == 2:
        return values[node]

    if maximizing:
        return max(
            minimax(depth + 1, node * 2, False, values),
            minimax(depth + 1, node * 2 + 1, False, values)
        )
    else:
        return min(
            minimax(depth + 1, node * 2, True, values),
            minimax(depth + 1, node * 2 + 1, True, values)
        )

values = [3, 5, 2, 9]
print("Best value:", minimax(0, 0, True, values))
```

```python
from collections import deque

def solve():
    q = deque([((3, 3, 1), [])])
    visited = set()

    while q:
        (m, c, b), path = q.popleft()

        if (m, c, b) == (0, 0, 0):
            print(path + [(m, c, b)])
            return

        if (m, c, b) in visited:
            continue
        visited.add((m, c, b))

        for x, y in [(1,0), (2,0), (0,1), (0,2), (1,1)]:
            nm, nc = (m-x, c-y) if b else (m+x, c+y)

            if 0 <= nm <= 3 and 0 <= nc <= 3:
                if (nm == 0 or nm >= nc) and (3-nm == 0 or 3-nm >= 3-nc):
                    q.append(((nm, nc, 1-b), path + [(m, c, b)]))

solve()
```

```python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

ratings = np.array([
    [5, 4, 0, 1],
    [4, 5, 0, 2],
    [1, 0, 5, 4],
    [0, 1, 4, 5]
])

similarity = cosine_similarity(ratings)

user = 0
scores = similarity[user] @ ratings
scores = scores / (similarity[user].sum() + 1e-9)

unrated = np.where(ratings[user] == 0)[0]
recommended = unrated[np.argmax(scores[unrated])]

print("Recommended Item:", recommended + 1)
```

```python
board = [" "] * 9

def show():
    for i in range(0, 9, 3):
        print(board[i], "|", board[i+1], "|", board[i+2])
        if i < 6:
            print("---------")

def winner(p):
    wins = [(0,1,2), (3,4,5), (6,7,8),
            (0,3,6), (1,4,7), (2,5,8),
            (0,4,8), (2,4,6)]
    return any(all(board[i] == p for i in w) for w in wins)

for turn in range(9):
    show()
    p = "X" if turn % 2 == 0 else "O"
    pos = int(input(f"Player {p}, enter position (1-9): ")) - 1

    if pos < 0 or pos > 8 or board[pos] != " ":
        print("Invalid position! Try again.")
        continue

    board[pos] = p

    if winner(p):
        show()
        print("Player", p, "wins!")
        break
else:
    show()
    print("Draw!")
```

```python
from collections import deque

def water_jug():
    q = deque([(0, 0)])
    visited = set()

    while q:
        a, b = q.popleft()

        if (a, b) in visited:
            continue
        visited.add((a, b))
        print(a, b)

        if a == 2 or b == 2:
            print("Goal Reached")
            return

        q.extend([
            (4, b), (a, 3), (0, b), (a, 0),
            (max(0, a-(3-b)), min(3, a+b)),
            (min(4, a+b), max(0, b-(4-a)))
        ])

water_jug()
```
